```python
#handle the outliers

#Outliers are data points that are significantly higher or lower than most values in a dataset.

#In the context of BMI, outliers are extremely low or high BMI values that differ from the majority of patients

#Outliers in BMI are extreme values that change the mean, not balanced distributions, affect graphs, and can
#mislead analysis, so they are often handled using median, trimming, or capping.
```

```python
import pandas as pd

# Example: BMI column from dataset
bmi = df['bmi']

# Calculate quartiles
Q1 = bmi.quantile(0.25)
Q3 = bmi.quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate outlier boundaries
lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

# Detect outliers
bmi_outliers = df[(bmi < lower_bound) | (bmi > upper_bound)]
print("Outliers:\n", bmi_outliers[['age','bmi']])
```

```python
#imbalance data In this dataset, stroke is the target variable, and all other columns (age, bmi, gender, etc.) are features used to predict stroke.
```

```python
#If a model is trained on highly imbalanced data, it may always predict the majority class (0), 
#ignoring the minority class (1). This means that no matter what input features you give, the model outputs 0 every time.
```

```python
df['stroke'].value_counts()
```

```python

df_majority = df[df['stroke'] == 0]
df_minority = df[df['stroke'] == 1]

```

```python
import pandas as pd
from sklearn.utils import resample

df_minority_upsampled = resample(
    df_minority,
    replace=True,               # sample with replacement
    n_samples=len(df_majority), # match majority class
    random_state=42
)
```

```python
df_balanced = pd.concat([df_majority, df_minority_upsampled])

# Check new class distribution
df_balanced['stroke'].value_counts()
```

```python
#synthetic=point1+fraction×(point2−point1)

#fraction=0.5

#[45,20]+0.5∗([50,30]−[45,20])=[45,20]+[2.5,5]=[47.5,25]


#| age | bmi | stroke |
#| --- | --- | ------ |
#| 45  | 20  | 1      |
#| 50  | 30  | 1      |


# | age  | bmi | stroke |                                        |
#| ---- | --- | ------ | -------------------------------------- |
#| 45   | 20  | 1      | ← original                             |
#| 50   | 30  | 1      | ← original                             |
#| 47.5 | 25  | 1      | ← synthetic (SMOTE, between 45 and 50) |
                            |
```

```python
from imblearn.over_sampling import SMOTE

# Features (numeric columns only) and target
X = df[['age', 'avg_glucose_level', 'bmi']]  # numeric features
y = df['stroke']

# Apply SMOTE
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)

# Check new class distribution
import pandas as pd
pd.Series(y_resampled).value_counts()
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

encoder = OneHotEncoder(drop='first')

encoded = encoder.fit_transform(
    df[['work_type', 'smoking_status']]
).toarray()

encoder_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(['work_type', 'smoking_status'])
)

encoder_df

df = pd.concat([df, encoder_df], axis=1)
```
